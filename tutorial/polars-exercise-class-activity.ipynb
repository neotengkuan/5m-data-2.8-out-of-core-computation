### Exercise 2.2: Complex Filtering

**Business Context**: PrimeTech Capital wants to identify high-volatility tech stocks.

**Tasks**:
1. Filter stock data where the daily price range (high - low) is greater than $10
2. Filter COVID data for dates between "2020-03-01" and "2020-06-01"
3. Filter sales data for orders with discount > 0.15 AND profit > 100

## PART 3: Grouping and Aggregation

### Exercise 3.1: Basic Grouping

**Business Context**: SuperMart wants to understand sales performance by different dimensions.

**Tasks**:
1. Group sales data by 'region' and calculate total sales
2. Group COVID data by 'country' and find maximum daily cases
3. Group stock data by 'symbol' and calculate average closing price
4. Count the number of records in each population dataset by year


```python
# 3.1 — Basic grouping
sales_by_region = sales_df.group_by("region").agg(
    pl.col("sales").sum().alias("total_sales")
)
print(sales_by_region)

covid_by_country = covid_df.group_by("country").agg(
    pl.col("new_cases").max().alias("max_daily_cases")
)
print(covid_by_country.head())

stocks_by_symbol = stocks_df.group_by("symbol").agg(
    pl.col("close").mean().alias("avg_closing_price")
)
print(stocks_by_symbol)

population_by_year = population_df.group_by("year").agg(
    pl.len().alias("record_count")
).sort("year")
print(population_by_year)
```

**Expected Functions**: Basic grouping and aggregation

### Exercise 3.2: Advanced Aggregations

**Business Context**: WHO needs comprehensive statistics for their pandemic response.

**Tasks**:
1. For each country in COVID data, calculate:
   - Total cases, maximum daily cases, average daily cases
   - Standard deviation of daily cases, median daily cases
2. For each stock symbol, calculate:
   - Mean closing price, min/max prices, price volatility (std dev)
3. For sales data by category, calculate:
   - Total sales, total profit, average discount, profit margin


```python
# 3.2 — Advanced aggregations
covid_stats = covid_df.group_by("country").agg([
    pl.col("new_cases").sum().alias("total_cases"),
    pl.col("new_cases").max().alias("max_daily_cases"),
    pl.col("new_cases").mean().alias("avg_daily_cases"),
    pl.col("new_cases").std().alias("std_daily_cases"),
    pl.col("new_cases").median().alias("median_daily_cases"),
])
print(covid_stats.head())

stock_stats = stocks_df.group_by("symbol").agg([
    pl.col("close").mean().alias("mean_close"),
    pl.col("close").min().alias("min_close"),
    pl.col("close").max().alias("max_close"),
    pl.col("close").std().alias("price_volatility"),
])
print(stock_stats)

sales_category_stats = sales_df.group_by("category").agg([
    pl.col("sales").sum().alias("total_sales"),
    pl.col("profit").sum().alias("total_profit"),
    pl.col("discount").mean().alias("avg_discount"),
]).with_columns(
    (pl.col("total_profit") / pl.col("total_sales") * 100).alias("profit_margin_pct")
)
print(sales_category_stats)
```

## PART 4: Time Series Analysis

### Exercise 4.1: Date Operations

**Business Context**: Analyzing trends over time requires proper date handling.

**Tasks**:
1. Convert date columns to datetime format in all relevant datasets
2. Extract year, month, and day of week from COVID data dates
3. Filter stock data for trading days in 2022 only
4. Calculate the number of days between order_date and ship_date in sales data


```python
# 4.1 — Date operations
# Parse dates. strict=False prevents a bad value from crashing the whole conversion.
covid_dates = covid_df.with_columns(
    pl.col("date").str.strptime(pl.Date, strict=False)
)
stocks_dates = stocks_df.with_columns(
    pl.col("date").str.strptime(pl.Date, strict=False)
)
sales_dates = sales_df.with_columns([
    pl.col("order_date").str.strptime(pl.Date, strict=False),
    pl.col("ship_date").str.strptime(pl.Date, strict=False),
])

covid_date_parts = covid_dates.with_columns([
    pl.col("date").dt.year().alias("year"),
    pl.col("date").dt.month().alias("month"),
    pl.col("date").dt.weekday().alias("day_of_week"),
])
print(covid_date_parts.head())

stocks_2022 = stocks_dates.filter(pl.col("date").dt.year() == 2022)
print(stocks_2022.head())

sales_shipping_days = sales_dates.with_columns(
    (pl.col("ship_date") - pl.col("order_date")).dt.total_days().alias("shipping_days")
)
print(sales_shipping_days.head())
```

**Expected Functions**: Date parsing and manipulation

### Exercise 4.2: Rolling Functions

**Business Context**: PrimeTech Capital wants to smooth out stock price volatility with moving averages.

**Tasks**:
1. Calculate 7-day rolling average of new COVID cases for each country
2. Calculate 30-day rolling average of stock closing prices for each symbol
3. Calculate 7-day rolling maximum and minimum for stock prices
4. Calculate 14-day rolling sum of sales by region


```python
# 4.2 — Rolling calculations
# Sort first so each rolling window follows time order.
covid_rolling = (
    covid_df.with_columns(pl.col("date").str.strptime(pl.Date, strict=False))
    .sort(["country", "date"])
    .with_columns(
        pl.col("new_cases").rolling_mean(window_size=7).over("country").alias("cases_7d_avg")
    )
)
print(covid_rolling.head())

stocks_rolling = (
    stocks_df.with_columns(pl.col("date").str.strptime(pl.Date, strict=False))
    .sort(["symbol", "date"])
    .with_columns([
        pl.col("close").rolling_mean(window_size=30).over("symbol").alias("close_30d_avg"),
        pl.col("high").rolling_max(window_size=7).over("symbol").alias("high_7d_max"),
        pl.col("low").rolling_min(window_size=7).over("symbol").alias("low_7d_min"),
    ])
)
print(stocks_rolling.head())

sales_rolling = (
    sales_df.with_columns(pl.col("order_date").str.strptime(pl.Date, strict=False))
    .sort(["region", "order_date"])
    .with_columns(
        pl.col("sales").rolling_sum(window_size=14).over("region").alias("sales_14row_sum")
    )
)
print(sales_rolling.head())
```

**Expected Functions**: Rolling calculations

## PART 5: Window Functions

### Exercise 5.1: Window Calculations

**Business Context**: Compare performance within groups without losing individual records.

**Tasks**:
1. For each country, calculate the percentage of total global cases
2. Rank stock symbols by their average closing price
3. Calculate each region's share of total sales
4. Find each customer's total spending and rank them


```python
# 5.1 — Window calculations
# 1) Each country's total cases as a percentage of the global total.
country_case_share = (
    covid_df.group_by("country")
    .agg(pl.col("new_cases").sum().alias("country_cases"))
    .with_columns(
        (pl.col("country_cases") / pl.col("country_cases").sum() * 100).alias("global_case_share_pct")
    )
)
print(country_case_share.head())

# 2) Rank symbols by average closing price.
stock_rank = (
    stocks_df.group_by("symbol")
    .agg(pl.col("close").mean().alias("avg_close"))
    .with_columns(pl.col("avg_close").rank(method="dense", descending=True).alias("rank"))
    .sort("rank")
)
print(stock_rank)

# 3) Each region's share of total sales.
region_share = (
    sales_df.group_by("region")
    .agg(pl.col("sales").sum().alias("region_sales"))
    .with_columns(
        (pl.col("region_sales") / pl.col("region_sales").sum() * 100).alias("sales_share_pct")
    )
)
print(region_share)

# 4) Customer total spending and rank.
customer_rank = (
    sales_df.with_columns(
        pl.col("sales").sum().over("customer_id").alias("customer_total_spending")
    )
    .select(["customer_id", "customer_total_spending"])
    .unique()
    .with_columns(
        pl.col("customer_total_spending").rank(method="dense", descending=True).alias("customer_rank")
    )
    .sort("customer_rank")
)
print(customer_rank.head())
```

## PART 6: Handling Missing Data

### Exercise 6.1: Missing Data Detection and Treatment

**Business Context**: Real-world data often has missing values that need handling.

**Tasks**:
1. Find all missing values in each dataset
2. Drop rows with any missing values from COVID data
3. Fill missing population values with the forward fill strategy
4. Replace missing values with the column mean for numeric columns


```python
# 6.1 — Missing data
# 1) Count nulls in every column of every dataset.
for name, df in datasets.items():
    print(f"\n=== Missing values: {name} ===")
    print(df.null_count())

# 2) Drop COVID rows containing any null.
covid_no_nulls = covid_df.drop_nulls()
print(covid_no_nulls.head())

# 3) Forward-fill missing population values within each country
#    after sorting by country and year.
population_filled = (
    population_df
    .sort(["country_name", "year"])
    .with_columns(
        pl.col("population")
        .fill_null(strategy="forward")
        .over("country_name")
    )
)
print(population_filled.head())

# 4) Replace nulls in numeric columns with each column's mean.
def fill_numeric_nulls_with_mean(df):
    numeric_cols = [
        name for name, dtype in df.schema.items()
        if dtype.is_numeric()
    ]
    return df.with_columns([
        pl.col(c).fill_null(pl.col(c).mean()).alias(c)
        for c in numeric_cols
    ])

sales_mean_filled = fill_numeric_nulls_with_mean(sales_df)
print(sales_mean_filled.head())
```

**Expected Functions**: Missing data handling

## PART 7: Creating New Columns

### Exercise 7.1: Column Transformations

**Business Context**: Create new metrics and calculated fields for analysis.

**Tasks**:
1. Create a 'daily_volatility' column for stocks (high - low) / close
2. Create a 'case_fatality_rate' column for COVID data (deaths/cases)
3. Create a 'profit_margin' column for sales data (profit/sales)
4. Add a row count column to identify each record


```python
# 7.1 — Create new columns
stocks_enriched = stocks_df.with_columns(
    ((pl.col("high") - pl.col("low")) / pl.col("close")).alias("daily_volatility")
)
print(stocks_enriched.head())

# Use total_cases/total_deaths if present; otherwise use cases/deaths if those are the dataset names.
cases_col = "total_cases" if "total_cases" in covid_df.columns else "cases"
deaths_col = "total_deaths" if "total_deaths" in covid_df.columns else "deaths"
covid_enriched = covid_df.with_columns(
    pl.when(pl.col(cases_col) > 0)
      .then(pl.col(deaths_col) / pl.col(cases_col) * 100)
      .otherwise(None)
      .alias("case_fatality_rate_pct")
)
print(covid_enriched.head())

sales_enriched = sales_df.with_columns(
    pl.when(pl.col("sales") != 0)
      .then(pl.col("profit") / pl.col("sales") * 100)
      .otherwise(None)
      .alias("profit_margin_pct")
)
print(sales_enriched.head())

# Add an explicit record number.
sales_numbered = sales_df.with_row_index("row_number", offset=1)
print(sales_numbered.head())
```

## PART 8: Reshaping Data

### Exercise 8.1: Pivoting and Melting

**Business Context**: Different analysis requires different data layouts.

**Tasks**:
1. Pivot COVID data to have countries as columns and dates as rows (for new_cases)
2. Melt stock data to have 'price_type' (open, high, low, close) and 'price' columns
3. Pivot sales data to show total sales by region and category
4. Concatenate all stock data with company information


```python
# 8.1 — Reshaping
# 1) Dates as rows, countries as columns, values = new_cases.
covid_pivot = covid_df.pivot(
    values="new_cases",
    index="date",
    on="country",
    aggregate_function="sum"
).sort("date")
print(covid_pivot.head())

# 2) Convert the four price columns from wide to long format.
stocks_long = stocks_df.unpivot(
    on=["open", "high", "low", "close"],
    index=[c for c in stocks_df.columns if c not in ["open", "high", "low", "close"]],
    variable_name="price_type",
    value_name="price"
)
print(stocks_long.head())

# 3) Total sales by region and category.
sales_pivot = sales_df.pivot(
    values="sales",
    index="region",
    on="category",
    aggregate_function="sum"
)
print(sales_pivot)

# 4) Company information belongs beside stock rows, so a join is appropriate.
# pl.concat() stacks/combines compatible frames; it does not match rows by symbol.
stocks_with_company = stocks_df.join(company_df, on="symbol", how="left")
print(stocks_with_company.head())
```

## PART 9: Joining Datasets

### Exercise 9.1: Data Joins

**Business Context**: Combine datasets to get comprehensive insights.

**Tasks**:
1. Inner join stock data with company information on 'symbol'
2. Left join sales data with a customer information table (you'll need to create a simple one)
3. Anti join to find stocks that don't have company information
4. Create a summary table by joining aggregated data from multiple sources


```python
# 9.1 — Joins
# 1) Inner join stocks to company information.
stocks_company_inner = stocks_df.join(company_df, on="symbol", how="inner")
print(stocks_company_inner.head())

# 2) Create a simple customer table, then left join it to sales.
customer_ids = sales_df.get_column("customer_id").unique().sort()
customer_info = pl.DataFrame({
    "customer_id": customer_ids,
    "customer_segment_label": [f"Customer {i+1}" for i in range(len(customer_ids))],
})

sales_with_customer = sales_df.join(customer_info, on="customer_id", how="left")
print(sales_with_customer.head())

# 3) Stocks whose symbol has no matching company record.
stocks_without_company = stocks_df.join(company_df, on="symbol", how="anti")
print(stocks_without_company)

# 4) Example summary: combine stock performance with company information.
stock_summary = stocks_df.group_by("symbol").agg([
    pl.col("close").mean().alias("avg_close"),
    pl.col("close").std().alias("close_volatility"),
    pl.len().alias("trading_records"),
])
company_summary = company_df.unique(subset=["symbol"])
combined_summary = stock_summary.join(company_summary, on="symbol", how="left")
print(combined_summary)
```

## PART 10: Advanced Analytics

### Exercise 10.1: Business Intelligence Queries

**Business Context**: Answer complex business questions that combine multiple techniques.

**Tasks**:

1. **COVID Impact Analysis**: Find the top 5 countries with the highest peak daily cases and their total death count

2. **Stock Performance Ranking**: Rank tech stocks by their return (latest price vs first price) and volatility

3. **Sales Trend Analysis**: Calculate month-over-month sales growth by region

4. **Market Opportunity Analysis**: Identify countries with growing populations but low technology adoption (you'll need to create a proxy metric)


```python
# 10.1 — Business intelligence queries
# 1) COVID impact: top 5 countries by peak daily cases + total deaths.
death_measure = "new_deaths" if "new_deaths" in covid_df.columns else (
    "total_deaths" if "total_deaths" in covid_df.columns else "deaths"
)
death_expr = (
    pl.col(death_measure).sum().alias("total_deaths")
    if death_measure == "new_deaths"
    else pl.col(death_measure).max().alias("total_deaths")
)

covid_top5 = (
    covid_df.group_by("country")
    .agg([pl.col("new_cases").max().alias("peak_daily_cases"), death_expr])
    .sort("peak_daily_cases", descending=True)
    .head(5)
)
print(covid_top5)

# 2) Stock performance: first/latest close, return and volatility.
stock_performance = (
    stocks_df.with_columns(pl.col("date").str.strptime(pl.Date, strict=False))
    .sort(["symbol", "date"])
    .group_by("symbol")
    .agg([
        pl.col("close").first().alias("first_close"),
        pl.col("close").last().alias("latest_close"),
        pl.col("close").std().alias("volatility"),
    ])
    .with_columns(
        ((pl.col("latest_close") / pl.col("first_close") - 1) * 100).alias("return_pct")
    )
    .with_columns([
        pl.col("return_pct").rank(method="dense", descending=True).alias("return_rank"),
        pl.col("volatility").rank(method="dense", descending=True).alias("volatility_rank"),
    ])
    .sort("return_rank")
)
print(stock_performance)

# 3) Month-over-month sales growth by region.
monthly_sales = (
    sales_df.with_columns(pl.col("order_date").str.strptime(pl.Date, strict=False))
    .with_columns(pl.col("order_date").dt.truncate("1mo").alias("month"))
    .group_by(["region", "month"])
    .agg(pl.col("sales").sum().alias("monthly_sales"))
    .sort(["region", "month"])
    .with_columns(
        (pl.col("monthly_sales").pct_change().over("region") * 100).alias("mom_growth_pct")
    )
)
print(monthly_sales.head())

# 4) Market opportunity: population growth + a simple technology-adoption proxy.
# The source does not define a real technology-adoption field, so we create an illustrative proxy.
pop_growth = (
    population_df.sort(["country_name", "year"])
    .group_by("country_name")
    .agg([
        pl.col("population").first().alias("first_population"),
        pl.col("population").last().alias("latest_population"),
    ])
    .with_columns(
        ((pl.col("latest_population") / pl.col("first_population") - 1) * 100).alias("population_growth_pct")
    )
)

# Illustrative proxy only: larger populations are assigned a lower adoption proxy score.
# Replace this with real internet/smartphone adoption data in a real analysis.
market_opportunity = (
    pop_growth.with_columns(
        (100 - pl.col("latest_population").rank(method="average", descending=False)
         / pl.len() * 100).alias("technology_adoption_proxy")
    )
    .filter(pl.col("population_growth_pct") > 0)
    .sort(["technology_adoption_proxy", "population_growth_pct"], descending=[False, True])
)
print(market_opportunity.head(10))
```
